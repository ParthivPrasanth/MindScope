```python
import pandas as pd
import numpy as np

# Load the empty file you saved earlier
df = pd.read_excel('/Users/siddharthanand21/mindscope_fusion/cleaned_responses.xlsx')

print("Columns before dropping:")
for i, col in enumerate(df.columns):
    print(f"  {i}: {col}")

# Drop by position — column 0 is Timestamp, column 1 is the consent statement
# This avoids any issues with matching the exact long column name strings
df = df.drop(df.columns[[0, 1]], axis=1)

print("Remaining columns:")
for i, col in enumerate(df.columns):
    print(f"  {i}: {col}")
print(f"\nTotal columns remaining: {len(df.columns)}")

# Map each column by position to a clean name
# This is safer than matching by string since long names can have hidden spaces

clean_names = [
    'age',
    'gender',
    'cgpa',
    'sleep_duration',
    'academic_pressure',
    'study_satisfaction',
    'financial_stress',
    'work_study_hours',
    'dietary_habits',
    'family_history',
    'suicidal_thoughts',
    'phq1',
    'phq2',
    'phq3',
    'phq4',
    'phq5',
    'phq6',
    'phq7',
    'phq8',
    'phq9',
    'free_text'
]

df.columns = clean_names

print("Columns after renaming:")
print(df.columns.tolist())

# Save this cleaned empty file with short column headers
# This is the file you open in Excel and fill in your synthetic data

df.to_excel('survey_data.xlsx', index=False)
print("Saved as survey_data.xlsx")
print("Open this file in Excel and fill in your 250 rows")
print("50 rows per severity level — one level per team member")
```
